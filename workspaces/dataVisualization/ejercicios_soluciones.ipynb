## Clase 1 — Introducción a Pandas

### Ejercicio
1. Agregar un nuevo estudiante.
2. Cambiar algunas calificaciones.
3. Encontrar el promedio del grupo y quién tiene la mejor calificación.

```python
import pandas as pd

# Crear el DataFrame original
datos = {
    "Nombre": ["Ana", "Luis", "Sofía", "Carlos", "Elena"],
    "Edad": [18, 17, 18, 19, 17],
    "Calificacion": [95, 88, 76, 90, 84],
    "Horas_estudio": [10, 7, 4, 8, 6]
}

df = pd.DataFrame(datos)
df
```

### 1. Agregar un nuevo estudiante

```python
# Simplemente agregamos a María en los datos originales
datos = {
    "Nombre": ["Ana", "Luis", "Sofía", "Carlos", "Elena", "María"],
    "Edad": [18, 17, 18, 19, 17, 18],
    "Calificacion": [95, 88, 76, 90, 84, 92],
    "Horas_estudio": [10, 7, 4, 8, 6, 9]
}

df = pd.DataFrame(datos)
df
```

#### Forma avanzada: usar `pd.concat()` para agregar sin reescribir todo

```python
# Con pd.concat() no necesitas reescribir todos los datos
# Solo creas el nuevo estudiante y lo unes al DataFrame existente
nuevo = pd.DataFrame([{
    "Nombre": "María",
    "Edad": 18,
    "Calificacion": 92,
    "Horas_estudio": 9
}])

df_avanzado = pd.concat([df, nuevo], ignore_index=True)
df_avanzado
```

### 2. Cambiar algunas calificaciones

```python
# Cambiar calificación de Luis y Elena
df["Calificacion"] = [95, 93, 76, 90, 78, 92]
df
```

#### Forma avanzada: usar `.loc[]` para cambiar valores individuales

```python
# Con .loc[] cambias solo el valor que necesitas, sin tocar los demás
# .loc[fila, columna] = nuevo valor
df.loc[1, "Calificacion"] = 93   # Luis
df.loc[4, "Calificacion"] = 78   # Elena
df
```

### 3. Encontrar el promedio del grupo y quién tiene la mejor calificación

```python
# Promedio del grupo
print("Promedio:", df["Calificacion"].mean())

# Mejor calificación
print("Mejor calificación:", df["Calificacion"].max())
print("Peor calificación:", df["Calificacion"].min())
```

#### Forma avanzada: usar `idxmax()` para encontrar el nombre directamente

```python
# idxmax() te da el índice (número de fila) del valor más alto
# Así puedes buscar el nombre de esa fila automáticamente
idx = df["Calificacion"].idxmax()
print("Mejor estudiante:", df.loc[idx, "Nombre"], "con", df.loc[idx, "Calificacion"])
```

## Clase 2 — Visualización de Datos con Matplotlib

### Ejercicio
1. Cambiar las horas de estudio.
2. Cambiar algunas calificaciones.
3. Crear nuevamente los tres gráficos.

Responder:
- ¿Quién estudió más?
- ¿Quién tiene la mejor calificación?
- ¿Existe relación entre estudiar más y sacar mejor resultado?

```python
import matplotlib.pyplot as plt

# Dataset con horas y calificaciones modificadas
datos2 = {
    "Nombre": ["Ana", "Luis", "Sofía", "Carlos", "Elena"],
    "Calificacion": [91, 85, 80, 95, 70],
    "Horas_estudio": [8, 6, 7, 11, 3]
}

df2 = pd.DataFrame(datos2)
df2
```

### Gráfico de barras

```python
plt.bar(df2["Nombre"], df2["Calificacion"])
plt.title("Calificaciones por estudiante")
plt.xlabel("Nombre")
plt.ylabel("Calificación")
plt.show()
```

### Gráfico de línea

```python
plt.plot(df2["Nombre"], df2["Horas_estudio"], marker="o")
plt.title("Horas de estudio")
plt.xlabel("Estudiantes")
plt.ylabel("Horas")
plt.show()
```

### Gráfico de dispersión

```python
plt.scatter(df2["Horas_estudio"], df2["Calificacion"])
plt.title("Relación estudio vs calificación")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificación")
plt.show()
```

### Respuestas

```python
# ¿Quién estudió más?
print("Máximo horas de estudio:", df2["Horas_estudio"].max())
# R: Carlos con 11 horas

# ¿Quién tiene la mejor calificación?
print("Mejor calificación:", df2["Calificacion"].max())
# R: Carlos con 95

# ¿Existe relación entre estudiar más y sacar mejor resultado?
# R: Sí, en el gráfico de dispersión se ve que
# los que estudian más horas tienden a tener mejor calificación.
```

#### Forma avanzada: usar `idxmax()` y `.corr()` para respuestas automáticas

```python
# idxmax() encuentra la fila con el valor más alto
# Así no tienes que buscar manualmente en la tabla
idx_estudio = df2["Horas_estudio"].idxmax()
print("Quien más estudió:", df2.loc[idx_estudio, "Nombre"], "con", df2.loc[idx_estudio, "Horas_estudio"], "horas")

idx_calif = df2["Calificacion"].idxmax()
print("Mejor calificación:", df2.loc[idx_calif, "Nombre"], "con", df2.loc[idx_calif, "Calificacion"])

# .corr() calcula la correlación entre dos columnas
# Un valor cercano a 1 significa relación positiva fuerte
correlacion = df2["Horas_estudio"].corr(df2["Calificacion"])
print("Correlación:", round(correlacion, 2))
```
