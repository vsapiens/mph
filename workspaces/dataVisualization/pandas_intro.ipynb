## 1. Importar Pandas

`pd` es un alias común para la librería.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Crear datos

Cada clave del diccionario representa una **columna**.

```python
datos = {
    "Nombre": [
        "Ana", "Luis", "Sofía", "Carlos", "Elena",
        "Diego", "Valentina", "Mateo", "Camila", "Andrés",
        "Isabella", "Sebastián"
    ],
    "Edad": [18, 17, 18, 19, 17, 18, 17, 19, 18, 17, 18, 19],
    "Calificacion": [95, 88, 76, 90, 84, 72, 91, 65, 78, 85, 93, 70],
    "Horas_estudio": [10, 7, 4, 8, 6, 3, 9, 2, 5, 7, 10, 3],
    "Asistencia": [98, 85, 70, 92, 80, 65, 95, 55, 75, 88, 100, 60],
    "Materia": [
        "Matemáticas", "Ciencias", "Historia", "Matemáticas", "Ciencias",
        "Historia", "Matemáticas", "Ciencias", "Historia", "Matemáticas",
        "Ciencias", "Historia"
    ]
}
```

## 3. Crear un DataFrame

Un **DataFrame** es una tabla similar a Excel.

```python
df = pd.DataFrame(datos)
print(df)
```

## 7. Filtrar datos

Esto devuelve solo los estudiantes con calificación mayor o igual a 85.

```python
aprobados = df[df["Calificacion"] >= 85]
print(aprobados)
```

## 8. Crear una nueva columna

```python
df["Resultado"] = df["Calificacion"].apply(
    lambda x: "Aprobado" if x >= 70 else "Reprobado"
)
df
```

## 9. Visualización de datos

Ahora que tenemos más información podemos representarla gráficamente para detectar patrones.

```python
# --- Gráfico 1: Calificaciones por estudiante (barras coloreadas por resultado) ---
colores = ["green" if r == "Aprobado" else "red" for r in df["Resultado"]]

plt.figure(figsize=(12, 4))
bars = plt.bar(df["Nombre"], df["Calificacion"], color=colores)
plt.axhline(y=70, color="gray", linestyle="--", label="Mínimo aprobatorio (70)")
plt.title("Calificación por estudiante")
plt.xlabel("Estudiante")
plt.ylabel("Calificación")
plt.xticks(rotation=30, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# --- Gráfico 2: Dispersión Horas de estudio vs Calificación ---
# El tamaño de cada punto representa el porcentaje de asistencia
sizes = df["Asistencia"] * 2

materias = df["Materia"].unique()
color_map = {"Matemáticas": "blue", "Ciencias": "orange", "Historia": "purple"}

plt.figure(figsize=(8, 5))
for materia in materias:
    mask = df["Materia"] == materia
    plt.scatter(
        df.loc[mask, "Horas_estudio"],
        df.loc[mask, "Calificacion"],
        s=sizes[mask],
        color=color_map[materia],
        alpha=0.7,
        label=materia
    )

for _, row in df.iterrows():
    plt.annotate(row["Nombre"], (row["Horas_estudio"], row["Calificacion"]),
                 textcoords="offset points", xytext=(5, 3), fontsize=7)

plt.title("Horas de estudio vs Calificación\n(tamaño del punto = % asistencia)")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificación")
plt.legend(title="Materia")
plt.tight_layout()
plt.show()
```

```python
# --- Gráfico 3: Promedio de calificación por materia (barras horizontales) ---
promedio_materia = df.groupby("Materia")["Calificacion"].mean().sort_values()

plt.figure(figsize=(7, 3))
plt.barh(promedio_materia.index, promedio_materia.values,
         color=["purple", "orange", "blue"])
plt.axvline(x=df["Calificacion"].mean(), color="gray", linestyle="--",
            label=f"Promedio general ({df['Calificacion'].mean():.1f})")
plt.title("Promedio de calificación por materia")
plt.xlabel("Calificación promedio")
plt.legend()
plt.tight_layout()
plt.show()
```
